```python
def display_game(game_list):
    print("Here is the current list: ")
    print(game_list)
```

```python
def position_choice():
    choice = 'wrong'
    
    while choice not in ['0', '1', '2']:
        choice = input("Please enter an index of choice (0-2)")
        
        if choice.isdigit() == False:
            print("Sorry, your choice is not a digit!")
            
        if choice not in ['0', '1', '2']:
            print("Sorry, your choice is not in range (0-2)")
        
    return int(choice)
```

```python
def replacement_choice(game_list, position_choice):
    replacement = input("Please enter a string to change the index")
    
    game_list[position_choice] = replacement
    
    return game_list
```

```python
def game_on():
    choice = False
    
    while choice not in ['Y', 'N']:
        choice = input("Continue playing? (Y or N)")
        
        if choice not in ['Y', 'N']:
            print("Sorry, don't understand, please type Y or N")

        if choice == 'Y':
            return True
        elif choice == 'N':
            return False
```

```python
game_list = [1, 2, 3]
game_continue = True

while game_continue == True:
    display_game(game_list)
    
    choice = position_choice()
    
    game_list = replacement_choice(game_list, choice)
    
    display_game(game_list)
    
    game_continue = game_on()
```
